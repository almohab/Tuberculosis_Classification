# VGG16 — 20 Epochs: Tuberculosis Chest X-ray Classification

Training and evaluation of a VGG16 transfer-learning model for binary tuberculosis classification from chest X-ray images.

**Dataset path expected:** `data/TB_Chest_Radiography_Database/`

This notebook preserves the existing cell outputs and results from the uploaded notebook. When you run it again, it will train and evaluate the model and save the model, metrics, classification report, and figures into the repository-relative `models/` and `results/` folders.

**Expected dataset folders:**
- `data/TB_Chest_Radiography_Database/Normal/`
- `data/TB_Chest_Radiography_Database/Tuberculosis/`

The test split is created by this notebook from the images found in those two class folders. Folder separation alone does not establish that duplicate or near-duplicate images are absent across splits; audit duplicates and patient overlap where metadata is available.

```python

# ============================================================
# VGG16 - TB CHEST X-RAY CLASSIFICATION
# 20 EPOCHS + COMPLETE EVALUATION
# ============================================================

import os
import time
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import VGG16
from tensorflow.keras.applications.vgg16 import preprocess_input
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout
from tensorflow.keras.optimizers import Adam

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

# ============================================================
# 1. SETTINGS
# ============================================================

SEED = 42
IMG_SIZE = 224
BATCH_SIZE = 32
EPOCHS = 20
THRESHOLD = 0.5

np.random.seed(SEED)
random.seed(SEED)
tf.random.set_seed(SEED)

categories = ['Normal', 'Tuberculosis']

# SAME DATASET PATH AS THE PREVIOUS CNN MODEL
# Repository-relative paths for GitHub sharing.
base_path = os.path.join("data", "TB_Chest_Radiography_Database")

# Keep generated artifacts organized in the repository.
os.makedirs("models", exist_ok=True)
os.makedirs(os.path.join("results", "figures"), exist_ok=True)
os.makedirs("results", exist_ok=True)

model_path = os.path.join("models", "VGG16_TB_20Epochs.keras")
results_file = os.path.join("results", "VGG16_20Epochs_Results.csv")
report_file = os.path.join("results", "VGG16_20Epochs_Classification_Report.csv")

training_plot_path = os.path.join("results", "figures", "VGG16_Training_History.png")
confusion_plot_path = os.path.join("results", "figures", "VGG16_Confusion_Matrix.png")
roc_plot_path = os.path.join("results", "figures", "VGG16_ROC_Curve.png")
pr_plot_path = os.path.join("results", "figures", "VGG16_PR_Curve.png")

print("=" * 60)
print("VGG16 TB CHEST X-RAY CLASSIFICATION")
print("=" * 60)

if not os.path.isdir(base_path):
    raise FileNotFoundError(
        f"Dataset directory not found:\n{base_path}\n"
        "Please check the dataset path."
    )

for category in categories:
    if not os.path.isdir(os.path.join(base_path, category)):
        raise FileNotFoundError(
            f"Missing class directory: "
            f"{os.path.join(base_path, category)}"
        )

# ============================================================
# 2. LOAD IMAGE PATHS AND LABELS
# ============================================================

image_paths = []
labels = []

valid_extensions = ('.png', '.jpg', '.jpeg', '.bmp')

for category in categories:
    folder_path = os.path.join(base_path, category)

    for filename in sorted(os.listdir(folder_path)):
        if filename.lower().endswith(valid_extensions):
            image_paths.append(
                os.path.join(folder_path, filename)
            )
            labels.append(categories.index(category))

df = pd.DataFrame({
    'image_path': image_paths,
    'label': labels
})

if df.empty:
    raise ValueError("No valid images were found in the dataset.")

if df['label'].nunique() != 2:
    raise ValueError(
        "Both Normal and Tuberculosis images are required."
    )

print("\nDATASET INFORMATION")
print("-" * 60)
print(f"Total images: {len(df)}")
print(df['label'].map({
    0: 'Normal',
    1: 'Tuberculosis'
}).value_counts())

# ============================================================
# 3. SPLIT DATA: 70% / 15% / 15%
# ============================================================

train_df, temp_df = train_test_split(
    df,
    test_size=0.30,
    random_state=SEED,
    stratify=df['label']
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=SEED,
    stratify=temp_df['label']
)

# Reset indices for consistent generator handling
train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("\nDATA SPLIT")
print("-" * 60)
print(f"Training images:   {len(train_df)}")
print(f"Validation images: {len(val_df)}")
print(f"Testing images:    {len(test_df)}")

# ============================================================
# 4. IMAGE GENERATORS
# ============================================================
# VGG16 pretrained on ImageNet expects 3-channel images.
# preprocess_input is used instead of rescale=1./255.
# Random augmentation is used ONLY for training.

train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

eval_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input
)

train_generator = train_datagen.flow_from_dataframe(
    dataframe=train_df,
    x_col='image_path',
    y_col='label',
    target_size=(IMG_SIZE, IMG_SIZE),
    color_mode='rgb',
    batch_size=BATCH_SIZE,
    class_mode='raw',
    shuffle=True,
    seed=SEED
)

val_generator = eval_datagen.flow_from_dataframe(
    dataframe=val_df,
    x_col='image_path',
    y_col='label',
    target_size=(IMG_SIZE, IMG_SIZE),
    color_mode='rgb',
    batch_size=BATCH_SIZE,
    class_mode='raw',
    shuffle=False
)

test_generator = eval_datagen.flow_from_dataframe(
    dataframe=test_df,
    x_col='image_path',
    y_col='label',
    target_size=(IMG_SIZE, IMG_SIZE),
    color_mode='rgb',
    batch_size=BATCH_SIZE,
    class_mode='raw',
    shuffle=False
)

# ============================================================
# 5. LOAD PRETRAINED VGG16
# ============================================================

print("\nLOADING VGG16")
print("-" * 60)

base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Freeze earlier layers; fine-tune the last four layers.
for layer in base_model.layers[:-4]:
    layer.trainable = False

for layer in base_model.layers[-4:]:
    layer.trainable = True

# ============================================================
# 6. BUILD MODEL
# ============================================================

model = Sequential([
    base_model,
    Flatten(),
    Dense(512, activation='relu'),
    Dropout(0.5),
    Dense(256, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()

# ============================================================
# 7. TRAIN MODEL - 20 EPOCHS
# ============================================================

print("\nTRAINING VGG16")
print("-" * 60)

training_start = time.perf_counter()

history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    verbose=1
)

training_time = time.perf_counter() - training_start

print(f"\nTraining time: {training_time:.2f} seconds")
print(f"Training time: {training_time / 60:.2f} minutes")

# ============================================================
# 8. TRAINING ACCURACY AND LOSS
# ============================================================

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('VGG16 Training and Validation Accuracy')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('VGG16 Training and Validation Loss')
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(training_plot_path, dpi=300)
plt.show()

# ============================================================
# 9. TEST SET EVALUATION
# ============================================================

print("\nTEST SET EVALUATION")
print("-" * 60)

test_loss, test_accuracy = model.evaluate(
    test_generator,
    verbose=1
)

print(f"Test loss: {test_loss:.4f}")
print(f"Test accuracy: {test_accuracy * 100:.2f}%")

# ============================================================
# 10. INFERENCE TIME
# ============================================================
# Warm up first. Then time the complete test set.
# Time includes model.predict processing and data feeding.

print("\nINFERENCE TIME")
print("-" * 60)

test_generator.reset()
warmup_images, _ = next(test_generator)
_ = model.predict(warmup_images[:min(4, len(warmup_images))],
                  verbose=0)

test_generator.reset()

inference_start = time.perf_counter()

y_pred_probs = model.predict(
    test_generator,
    verbose=1
).ravel()

total_inference_time = time.perf_counter() - inference_start

num_test_images = len(test_df)

inference_time_per_image = (
    total_inference_time / num_test_images
)

images_per_second = (
    num_test_images / total_inference_time
    if total_inference_time > 0 else 0
)

print(f"Test images: {num_test_images}")
print(f"Total inference time: {total_inference_time:.4f} seconds")
print(f"Time per image: {inference_time_per_image:.6f} seconds")
print(f"Time per image: {inference_time_per_image * 1000:.4f} ms")
print(f"Images per second: {images_per_second:.2f}")

# ============================================================
# 11. PREDICTIONS AND CONFUSION MATRIX
# ============================================================

y_true = test_df['label'].to_numpy(dtype=int)
y_pred = (y_pred_probs >= THRESHOLD).astype(int)

cm = confusion_matrix(y_true, y_pred, labels=[0, 1])

TN, FP, FN, TP = cm.ravel()

print("\nCONFUSION MATRIX")
print("-" * 60)
print("Rows = actual class; columns = predicted class")
print(cm)
print(f"True Negative (TN):  {TN}")
print(f"False Positive (FP): {FP}")
print(f"False Negative (FN): {FN}")
print(f"True Positive (TP):  {TP}")

plt.figure(figsize=(7, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=categories,
    yticklabels=categories
)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('VGG16 Confusion Matrix')
plt.tight_layout()
plt.savefig(confusion_plot_path, dpi=300)
plt.show()

# ============================================================
# 12. SENSITIVITY, SPECIFICITY, PRECISION, NPV
# ============================================================

sensitivity = (
    TP / (TP + FN) if (TP + FN) > 0 else 0.0
)

specificity = (
    TN / (TN + FP) if (TN + FP) > 0 else 0.0
)

precision_value = (
    TP / (TP + FP) if (TP + FP) > 0 else 0.0
)

npv = (
    TN / (TN + FN) if (TN + FN) > 0 else 0.0
)

accuracy_value = accuracy_score(y_true, y_pred)

print("\nCLINICAL PERFORMANCE METRICS")
print("-" * 60)
print(f"Accuracy:    {accuracy_value * 100:.2f}%")
print(f"Sensitivity: {sensitivity * 100:.2f}%")
print(f"Specificity: {specificity * 100:.2f}%")
print(f"Precision:   {precision_value * 100:.2f}%")
print(f"NPV:         {npv * 100:.2f}%")

# ============================================================
# 13. CLASSIFICATION REPORT
# ============================================================

report = classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=categories,
    digits=4,
    zero_division=0,
    output_dict=True
)

print("\nCLASSIFICATION REPORT")
print("-" * 60)

print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1],
    target_names=categories,
    digits=4,
    zero_division=0
))

pd.DataFrame(report).transpose().to_csv(report_file)
print(f"Classification report saved: {report_file}")

# ============================================================
# 14. ROC CURVE AND ROC-AUC
# ============================================================

fpr, tpr, roc_thresholds = roc_curve(
    y_true,
    y_pred_probs
)

roc_auc = roc_auc_score(y_true, y_pred_probs)

plt.figure(figsize=(8, 6))
plt.plot(
    fpr,
    tpr,
    label=f'VGG16 ROC-AUC = {roc_auc:.4f}'
)
plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('VGG16 Receiver Operating Characteristic')
plt.legend(loc='lower right')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(roc_plot_path, dpi=300)
plt.show()

print(f"ROC-AUC: {roc_auc:.4f}")

# ============================================================
# 15. PRECISION-RECALL CURVE AND PR-AUC
# ============================================================

pr_precision, pr_recall, pr_thresholds = (
    precision_recall_curve(y_true, y_pred_probs)
)

pr_auc = auc(pr_recall, pr_precision)

average_precision = average_precision_score(
    y_true,
    y_pred_probs
)

plt.figure(figsize=(8, 6))
plt.plot(
    pr_recall,
    pr_precision,
    label=f'VGG16 PR-AUC = {pr_auc:.4f}'
)
plt.xlabel('Recall (Sensitivity)')
plt.ylabel('Precision')
plt.title('VGG16 Precision-Recall Curve')
plt.legend(loc='lower left')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(pr_plot_path, dpi=300)
plt.show()

print(f"PR-AUC: {pr_auc:.4f}")
print(f"Average Precision: {average_precision:.4f}")

# ============================================================
# 16. MODEL PARAMETERS
# ============================================================

total_params = model.count_params()

trainable_params = int(sum(
    np.prod(v.shape) for v in model.trainable_weights
))

non_trainable_params = int(sum(
    np.prod(v.shape) for v in model.non_trainable_weights
))

print("\nMODEL PARAMETERS")
print("-" * 60)
print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")
print(f"Non-trainable parameters: {non_trainable_params:,}")

# ============================================================
# 17. SAVE MODEL AND CALCULATE MODEL SIZE
# ============================================================

model.save(model_path)

model_size_bytes = os.path.getsize(model_path)
model_size_mb = model_size_bytes / (1024 ** 2)
model_size_kb = model_size_bytes / 1024

print("\nMODEL INFORMATION")
print("-" * 60)
print(f"Saved model: {model_path}")
print(f"Model size: {model_size_mb:.2f} MB")
print(f"Model size: {model_size_kb:.2f} KB")

# ============================================================
# 18. FINAL RESULTS TABLE
# ============================================================

results = pd.DataFrame({
    'Metric': [
        'Model',
        'Epochs',
        'Input Size',
        'Training Images',
        'Validation Images',
        'Testing Images',
        'Test Accuracy (%)',
        'Test Loss',
        'Sensitivity (%)',
        'Specificity (%)',
        'Precision (%)',
        'Negative Predictive Value (%)',
        'F1-Score (TB)',
        'ROC-AUC',
        'PR-AUC',
        'Average Precision',
        'True Negative (TN)',
        'False Positive (FP)',
        'False Negative (FN)',
        'True Positive (TP)',
        'Total Parameters',
        'Trainable Parameters',
        'Non-trainable Parameters',
        'Model Size (MB)',
        'Total Inference Time (s)',
        'Inference Time/Image (ms)',
        'Inference Time/Image (s)',
        'Images/Second',
        'Training Time (s)',
        'Training Time (min)'
    ],
    'Value': [
        'VGG16',
        EPOCHS,
        f'{IMG_SIZE} x {IMG_SIZE} x 3',
        len(train_df),
        len(val_df),
        len(test_df),
        f'{accuracy_value * 100:.2f}',
        f'{test_loss:.4f}',
        f'{sensitivity * 100:.2f}',
        f'{specificity * 100:.2f}',
        f'{precision_value * 100:.2f}',
        f'{npv * 100:.2f}',
        f'{f1_score(y_true, y_pred, zero_division=0):.4f}',
        f'{roc_auc:.4f}',
        f'{pr_auc:.4f}',
        f'{average_precision:.4f}',
        TN,
        FP,
        FN,
        TP,
        total_params,
        trainable_params,
        non_trainable_params,
        f'{model_size_mb:.2f}',
        f'{total_inference_time:.4f}',
        f'{inference_time_per_image * 1000:.4f}',
        f'{inference_time_per_image:.6f}',
        f'{images_per_second:.2f}',
        f'{training_time:.2f}',
        f'{training_time / 60:.2f}'
    ]
})

print("\nFINAL VGG16 RESULTS")
print("=" * 60)
print(results.to_string(index=False))

results.to_csv(results_file, index=False)

print("\nFILES SAVED")
print("-" * 60)
print(f"Model: {model_path}")
print(f"Results: {results_file}")
print(f"Classification report: {report_file}")
print(f"Plots: {training_plot_path}")
print(f"       {confusion_plot_path}")
print(f"       {roc_plot_path}")
print(f"       {pr_plot_path}")
print("\nVGG16 evaluation completed successfully.")
```
